```python
import numpy as np
```

NUMPY INDEXING AND SLICING OF VECTORS

```python
arr = np.arange(11,21)
arr
```

```python
arr_slice = arr[3::2]
arr_slice
```

```python
arr[4]
```

NUMPY INDEXING AND SLICING OF MATRIX

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[5]
```

```python
arr[0,0]
```

```python
slice = arr[0:2,1:3]
slice
```

```python
slice = arr[3:,3:]
slice
```

```python
arr[:,2]
```

BOOLEAN INDEXING

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr%2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
